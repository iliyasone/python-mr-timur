```python
# 16

from functools import lru_cache

@lru_cache(1000)
def g(n):
    if n <= 9:
        return n * 3
    elif n > 9:
        return g(n - 4) + 2

def f(n):
    a = g(n - 1) + g(n - 3)
    return a

for i in range(1, 42999):
    g(i)

print(f(42999))
print('hi')
```
