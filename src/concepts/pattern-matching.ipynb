```python
def match_error(body):
    match body:
        case (str() as text) | {'errors' : [{'message': str() as text}]} if "Rate limit exceeded" in text.strip() :
            return True
        case _:
            return False
```

```python
match_error({'errors' : [{'message': "Rate limit exceeded. gdsg"}]})
```
