## демо 2022

```python

WIN = 666

def moves(s: int):
    return (s + 3, s * 3, s + s**2)

def is_position_winning(s: int, turns: int = 1) -> bool:
    if turns == 0:
        return None
     
    if any(m >= WIN for m in moves(s)):
        return True
    
    if all(is_position_winning(m, turns - 1) is True for m in moves(s)):
        return False
    
    if any(is_position_winning(m, turns - 1) is False for m in moves(s)):
        return True
    
```

```python
for i in range(1, 30):
    print(i,'\t', is_position_winning(i, 4))
```

## демо 2025

```python

WIN = 19

def moves(s: int) -> int:
    return s-2, s-5, s//3


def is_position_winning(s: int, turns: int = 1) -> bool:
    if turns == 0:
        return None
     
    if any(m <= WIN for m in moves(s)):
        return True
    
    if all(is_position_winning(m, turns - 1) for m in moves(s)):
        return False
    
    if any(not is_position_winning(m, turns - 1) for m in moves(s) if is_position_winning(m, turns - 1) is not None):
        return True
    
```

```python
for i in range(WIN+1, 100):
    print(i,'\t', is_position_winning(i, 4), '\t',is_position_winning(i, 3))
```

## демо 2021

```python
WIN = 77
S1 = 7
S2_MAX = 69

def moves(s: tuple[int, int]):
    s1, s2 = s
    return (s1+1, s2), (s1, s2+1), (s1*2, s2), (s1, s2*2)

def is_position_winning(s: tuple[int, int], turns: int = 1) -> bool:
    if turns == 0:
        return None
     
    if any(sum(m) >= WIN for m in moves(s)):
        return True
    
    if all(is_position_winning(m, turns - 1) for m in moves(s)):
        return False
    
    if any(not is_position_winning(m, turns - 1) for m in moves(s) if is_position_winning(m, turns - 1) is not None):
        return True
    

def is_position_winning_bad_turn(s: tuple[int, int], turns: int = 1) -> bool:
    if turns == 0:
        return None
     
    if any(sum(m) >= WIN for m in moves(s)):
        return True
    
    if any(is_position_winning(m, turns - 1) for m in moves(s)):
        return False
    
    if any(not is_position_winning(m, turns - 1) for m in moves(s) if is_position_winning(m, turns - 1) is not None):
        return True
```

```python
for s2 in range(S2_MAX):
    print((S1, s2), is_position_winning_bad_turn((S1, s2), 2), is_position_winning((S1, s2), 1), sep='\t')
```
